```python
# imports
from matplotlib import pyplot as plt
import numpy as np
import sympy as sp
import torch
import os
import arviz as az
import pyro
import sys
sys.path.append(os.path.abspath(os.path.join('..')))

import post_hoc_inference
from post_hoc_inference import PostHocInference
from post_hoc_inference.core.pyro_backend import utils as pyro_utils
from post_hoc_inference.visualisation import predictive_plotting as pred_plot
```

```python
# create data
N_train = 20
N_test = 1000
seed = 42

noise_sigma = 1.  
np.random.seed(seed)
torch.manual_seed(seed)

def synthetic_function_1d(x, noise_sigma=1.):
    y = 2 * torch.sin(x) + 3
    y_obs = y + torch.randn(x.shape if len(x) > 0 else (1,)) * noise_sigma
    return y, y_obs


x_train = torch.linspace(0, 3*np.pi/2, steps=N_train)
x_test = torch.linspace(0, 4*np.pi/2, steps=N_test)

y_train, y_obs_train = synthetic_function_1d(x_train, noise_sigma)
y_test, y_obs_test = synthetic_function_1d(x_test, noise_sigma)

plt.scatter(x_train, y_obs_train, marker="x", color="black")
plt.plot(x_test, y_test, label="True Function")
plt.xlabel("x")
plt.ylabel("y")
```

```python
# Do post-hoc Inference on the given Sympy equation string.

# define sympy expression
x0 = sp.Symbol('x0')

expr_sp_true = 2. * sp.sin(x0) + 3.
expr_sp_quadratic = 2. * x0**2 - 1.1 * x0 + 5.

kernel = "nuts"

print("Ground truth")
print(sp.sympify(expr_sp_true))
print("Quadratic Function")
print(sp.sympify(expr_sp_quadratic))

post_hoc_inf = PostHocInference(
    expr=expr_sp_quadratic,
    input_symbols=[x0],
    output_dim=1,
    inference_params={"kernel_name": f"{kernel}", "n_particles":10, "vi_iter": 3000, "n_warmup_samples":10, "n_samples":10}
)


post_hoc_inf.fit(x_train.view((-1,1)), y_obs_train.view((-1,1)))
```

```python
post_hoc_inf.predict(x_train.view((-1,1)), sample_prior=False, parallel=True)["mu"].shape
```

```python
# Evaluation
os.makedirs(f"{post_hoc_inference.PROJECT_ROOT}/output", exist_ok=True)

post_hoc_inf.inference_diagnostics()

idata = post_hoc_inf.get_posterior()
if type(idata) is not az.InferenceData:
    idata = pyro_utils.guide_to_inference_data(post_hoc_inf._pyro_model, post_hoc_inf._guide, x_train.view((-1,1)), num_samples=3000)

az.plot_trace(idata)
plt.tight_layout()
plt.savefig(f"{post_hoc_inference.PROJECT_ROOT}/output/posterior_{kernel}.png")
# 
```
